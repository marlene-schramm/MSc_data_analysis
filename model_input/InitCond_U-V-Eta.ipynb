# Initial Conditions for Velocities and Eta

```python
import numpy as np
import pandas as pd
import xarray as xr
import xmitgcm.utils
import gsw as gsw
import matplotlib.pyplot as plt
```

## Velocities

```python
Nx = 50 # east-west
Ny = 100 # north-south
Nz = 50

v = 0.
 
dir_binaries = "../../MITgcm/so_plumes/input/"
prefix = "init_"
suffix = f"_Nx{Nx}_Ny{Ny}_Nz{Nz}"

velocities = np.full((Nx,Ny,Nz), v)

filename = prefix + f"U{int(v)}" + suffix + ".bin"
xmitgcm.utils.write_to_binary(velocities.flatten(order='F'), dir_binaries+filename)

filename = prefix + f"V{int(v)}" + suffix + ".bin"
xmitgcm.utils.write_to_binary(velocities.flatten(order='F'), dir_binaries+filename)

print(filename)
```

## Eta

```python
Nx = 50 # east-west
Ny = 100 # north-south

eta = 0.
 
dir_binaries = "../../MITgcm/so_plumes/input/"
prefix = "init_"
suffix = f"_Nx{Nx}_Ny{Ny}"

etas = np.full((Nx,Ny), eta)

filename = prefix + f"Eta{int(eta)}" + suffix + ".bin"
xmitgcm.utils.write_to_binary(etas.flatten(order='F'), dir_binaries+filename)

print(filename)
```

# Compare Temperature / Heat Content

In this Notebook, I want to make a plot that compares the final temperature profile/heat content for several MITgcm simulations.

```python
# packages for data processing
import numpy as np
import pandas as pd
import xarray as xr
from xmitgcm import open_mdsdataset
import gsw as gsw

# packages for plotting
import matplotlib.pyplot as plt
import plotly.graph_objects as go # interactive/3D plotting
import plotly.express as px
import cmocean.cm as cmo # ocean colormaps

# my own plotting functions
from plotting_functions import *

# technical packages
import warnings
```

We loop through the model runs. For each model run, we load the data, retrieve the data that we want to keep and discard the rest.

```python

model_runs = [f"MIZ{i:03d}" for i in range(9,16)] # model runs to loop through
delta_t = 4 # time step in seconds
output_dir = "../../MITgcm/so_plumes/"

final_avg_T_wholedomain = pd.DataFrame() 
final_avg_T_openwater = pd.DataFrame() 

# surpress the xmitgcm warning about future changes with timedelta
with warnings.catch_warnings():
    warnings.simplefilter("ignore", category=FutureWarning)

    for i in range(len(model_runs)):
        run = model_runs[i]
        
        # open dataset
        ds = open_mdsdataset(output_dir+run, prefix = ['Eta', 'U', 'W', 'T', 'V', 'S', 'PH'], delta_t = delta_t, geometry = "cartesian")
        # openwater subdomain
        ds_openwater = ds.isel(YC = slice(0,50))
        # select last time step
        ds_end = ds.isel(time = -1)
        ds_openwater_end = ds_openwater.isel(time=-1)

        if i == 0:
            # set depth as index
            final_avg_T_wholedomain.index = ds.Z
            final_avg_T_wholedomain.index.name = "depth"
            final_avg_T_openwater.index = ds.Z
            final_avg_T_openwater.index.name = "depth"
            
            # average T profile from initial conditions
            final_avg_T_wholedomain["initCond"] = ds_openwater.isel(time=0).T.mean(dim = ["YC","XC"])
            final_avg_T_openwater["initCond"] = ds_openwater.isel(time=0).T.mean(dim = ["YC","XC"])
            
        # calculate mean final temperature profile
        final_avg_T_wholedomain[run] = ds_end.T.mean(dim = ["YC","XC"])
        final_avg_T_openwater[run] = ds_openwater_end.T.mean(dim = ["YC","XC"])
        
final_avg_T_wholedomain.head()
```

```python
# whole domain

fig, axs = plt.subplots(2,3, figsize = (25,8))

# freezing point variations
axs[0].set_title("Freezing Point Variation")
axs[0].plot(final_avg_T_wholedomain.MIZ014, final_avg_T_wholedomain.index, label = "T_f = -1.9")
axs[0].plot(final_avg_T_wholedomain.MIZ009, final_avg_T_wholedomain.index, label = "Ref Run T_f = -2.0")
axs[0].plot(final_avg_T_wholedomain.MIZ015, final_avg_T_wholedomain.index, label = "T_f = -2.4")
axs[0].set_ylabel("Depth (m)")

# heat flux variations
axs[1].set_title("Heat Forcing Variation")
axs[1].plot(final_avg_T_wholedomain.MIZ010, final_avg_T_wholedomain.index, label = "Q = -250W/m²s")
axs[1].plot(final_avg_T_wholedomain.MIZ009, final_avg_T_wholedomain.index, label = "Ref Run Q = -500/m²s")
axs[1].plot(final_avg_T_wholedomain.MIZ011, final_avg_T_wholedomain.index, label = "Q = -1000/m²s")

# salt forcing variation
axs[2].set_title("Salt Forcing Variation")
axs[2].plot(final_avg_T_wholedomain.MIZ012, final_avg_T_wholedomain.index, label = "F_S = -0.004g/m²s")
axs[2].plot(final_avg_T_wholedomain.MIZ009, final_avg_T_wholedomain.index, label = "Ref Run F_S = -0.008g/m²s")
axs[2].plot(final_avg_T_wholedomain.MIZ013, final_avg_T_wholedomain.index, label = "F_S = -0.016g/m²s")

for ax in axs:
    ax.legend()
    ax.set_xlabel("T (°C)")

plt.suptitle("Final Average Temperute Profile over Whole Domain after 48h")
```
